```python
from matplotlib import pyplot as plt
import numpy as np

LEVELS = ["No Damage", "Minor", "Major", "Destroyed"]
PIXEL_RATIOS: dict[str, list[float]] = {
    "xBD"               : [76.35, 08.63, 08.99, 06.02],
    "Earthquake Turkey" : [95.89, 01.66, 01.48, 00.97],
    "Pakistan Flooding" : [79.82, 10.09, 02.91, 07.18],
    "Hurricane IDA"     : [88.84, 07.79, 02.96, 00.41],
}

for key, val in PIXEL_RATIOS.items():
    plt.bar(LEVELS, val)
    plt.ylim([0, 101])
    plt.title(key, fontweight="bold")
    plt.xlabel("Damage Level", fontweight="bold")
    plt.ylabel("Pixel Ratio", fontweight="bold")
    for i, percent in enumerate(val):
        plt.text(i, percent + 0.75, f"{percent}%", ha="center")
    plt.show()
```
